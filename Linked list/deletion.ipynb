```python
class Node:
    def __init__(self,val):
        self.val = val
        self.Next = None
        self.prev = None
      
```

```python
def insert_at_head(self,val):
    new_node = Node(val)
    if self.head is None:
        new_node = self.head
        new_node.prev = None
    else:
        new_node.next = self.head
        self.head.prev = new_node
        self.head = new_node 
```

```python
def append(self,val):
    new_node = Node(val)
    if self.head is None:
        new_node = self.head
    else:
        current = self.head
        while current is not None:
            current = current.next
        current.next=new_node
        new_node= current
        
```
